```python
#Create a function using *args that accepts any number of values and returns their total
def fun1(*args):
    return sum(args)

total = fun1(1,2,3,4,5)
print(f'Total of the given numbers is {total}')
```

```python
#Create a function using *args that returns the largest supplied number
def fun1(*args):
    l1 = list(args)

    max = l1[0]

    for i in range(len(l1)):

        if l1[i] > max:
            max = l1[i]

    return max

lar = fun1(1,2,3,4,5)
print(f'The largest supplied number in the given list is {lar}')
```
